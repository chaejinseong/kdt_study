# 사이킷런(Scikit-learn)
- 파이썬 기반의 대표적인 오픈소스 머신러닝 라이브러리로, 데이터 분석 및 예측 모델 구축에 광범위하게 활용됩니다. 
일관되고 직관적인 API 인터페이스를 바탕으로 지도학습(Supervised Learning)과 비지도학습(Unsupervised Learning) 알고리즘을 두루 지원합니다.
- 주요 기능으로는 분류(Classification), 회귀(Regression), 클러스터링(Clustering), 차원 축소(Dimensionality Reduction), 모델 평가 및 선택(Model Selection), 데이터 전처리(Preprocessing) 등이 있습니다. 수치 연산 효율성이 뛰어나고 머신러닝 워크플로우 전반을 손쉽게 구현할 수 있어, 
데이터 과학 및 AI 프로젝트에서 핵심적인 표준 라이브러리로 평가받고 있습니다.

## scikit-learn을 구성하는 주요 모듈
| 모듈명                     | 설명                                                    |
|---------------------------|---------------------------------------------------------|
| `sklearn.datasets`        | 내장된 예제 데이터 세트                                    |
| `sklearn.preprocessing`        | 다양한 데이터 전처리 기능 제공(변환, 정규화, 스케일링 등)                                   |
| `sklearn.feature_selection`        | 특징(feature)를 선택할 수 있는 기능 제공                                   |
| `sklearn.feature_extraction` | 특징(feature) 추출에 사용                    |
| `sklearn.decomposition` | 차원축소 관련 알고리즘 지원(PCA, NMF, Truncated SVD 등)                    |
| `sklearn.model_selection` | 교차 검증을 위해 데이터를 학습/테스트용으로 분리, 최적 파라미터를 추출하는 API 제공(GridSearch 등)                    |
| `sklearn.metrics`         | 분류, 회귀, 클러스터링, Pariwise에 대한 다양한 선능 측정 방법 제공(Accuracy, Precision, Recall, ROC-AUC, RMSE 등)                                |
| `sklearn.pipeline`         | 특정 처리 등의 변환과 ML 알고리즘 학습, 예측 등을 묶어서 실행할 수 있는 유틸리티 제공                                |
| `sklearn.linear_model`    | 선형 회귀, 릿지(Ridge), 라쏘(Lasso), 로지스틱 회귀 등 회귀 관련 알고리즘과 SGD(Stochastic Gradient Descent) 알고리즘 제공                                  |
| `sklearn.svm`    | 서포트 벡터 머신 알고리즘 제공                                  |
| `sklearn.neighbors`       | 최근접 이웃 알고리즘 (k-NN 등)                        |
| `sklearn.naive_bayes`       | 나비브 베이즈 알고리즘 제공(가우시안 NB, 다항 분포 NB 등)                        |
| `sklearn.tree`            | 의사 결정 트리 알고리즘 제공                                     |
| `sklearn.ensemble`        | 앙상블 알고리즘 제공 (Random Forest, AdaBoost, GradientBoost 등) |
| `sklearn.cluster`         | 비지도 학습 클러스터링 알고리즘 제공 (k-Means, 계층형 클러스터링, DBSCAN 등)                        |

## Iris 데이터셋
아이리스(Iris, 붓꽃) 데이터셋은 머신러닝 및 통계학 분야에서 가장 널리 활용되는 대표적인 벤치마크 데이터셋입니다. 
세 가지 붓꽃 품종인 세토사(Setosa), 버시컬러(Versicolor), 버지니카(Virginica)에 대한 관측 정보를 담고 있습니다. 
각 품종별로 꽃받침(Sepal)의 길이와 너비, 꽃잎(Petal)의 길이와 너비라는 4개의 특성(Features)으로 구성되며, 총 150개의 샘플 데이터가 수록되어 있습니다. 
품종별로 50개씩 균등하게 분포되어 있어, 다중 클래스 분류(Multi-class Classification) 문제를 학습하고 검증하기에 매우 적합합니다.

## 데이터셋 (Dataset)이란?
데이터셋은 머신러닝 및 데이터 과학에서 모델을 학습, 검증, 평가하기 위해 사용되는 데이터의 집합입니다. 
일반적으로 입력 변수에 해당하는 특성(Features)과 정답에 해당하는 레이블(Labels)로 구성되며, 
목적에 따라 학습용(Training Dataset), 검증용(Validation Dataset), 테스트용(Test Dataset) 데이터셋으로 분할하여 활용합니다. 
데이터셋의 품질과 규모는 모델의 최종 성패를 결정짓는 핵심 요인이므로, 적절한 데이터 전처리(Preprocessing)와 
특성 엔지니어링(Feature Engineering) 과정이 필수적입니다.

### 1. 서포트 벡터 머신 (SVM, Support Vector Machine)
- 개념: 지도학습(Supervised Learning) 기반의 대표적인 머신러닝 알고리즘으로, 주로 분류(Classification) 문제 해결에 활용되며 회귀(SVR) 분석에도 적용 가능합니다.

## 4단계, 모델 평가
### 혼동 행렬 (Confusion Matrix)
- 개념: 분류 모델의 예측 결과와 실제 정답 레이블 간의 관계를 시각화하고 요약하여 모델의 성능을 정밀하게 평가하는 도구입니다.
- 구성 요소: 이진 분류(Binary Classification) 기준으로 다음 네 가지 기본 수치로 이루어집니다.
   - True Positive (TP): 실제 양성(Positive)을 양성으로 올바르게 예측한 경우
   - True Negative (TN): 실제 음성(Negative)을 음성으로 올바르게 예측한 경우
   - False Positive (FP): 실제 음성을 양성으로 잘못 예측한 경우 (1종 오류)
   - False Negative (FN): 실제 양성을 음성으로 잘못 예측한 경우 (2종 오류)
- 활용 지표: 정답을 맞춘 TP와 TN, 오답인 FP와 FN의 조합을 통해 정확도(Accuracy), 정밀도(Precision), 재현율(Recall), F1 점수(F1-score) 등의 다각적 성능 지표를 산출합니다.
- 분석적 의의: 단순히 전체 정확도만으로 파악하기 어려운 클래스 불균형(Class Imbalance) 데이터셋에서 모델이 범하는 오차의 세부 유형을 파악하는 데 매우 유용합니다.
### 정확도(accuracy)
- 1번과, 4번 맞춘 것의 비율
- (TP + TN) / (TP + TN + FN + FP)

### 정밀도(presicion)
- positive 중에서 재대로 맞춘 것
- TP / (TP + FP)

### *재현률(recall)
- 실제 positive 중에서 positive를 맞춘 확률
- TP / (TP + FN)

### 오류률(Error Rate)
- 전체 데이터 중에서 잘못 분류한 비율
- 1 - accuracy
