## boolean selection, boolean indexing
- array 인덱싱 시, bool타입의 조건에 맞는 list를 전달하여 조건식을 만족하는 True의 결과물만 필터링한다.

```python
arr4 = np.random.randint(1, 100, size=20)
print(arr4)
```

```python
even_mask = arr4 % 5 == 0
arr4[even_mask]
```

```python
is_bound = arr4>30
arr4[is_bound]
```

## 다중 조건식
- & : and 조건식
- | : or 조건식

```python
# 홀수이면서 50보다 작은 값
# and:
np.random.seed(2026)
arr5 : np.random.randint(100, size = 15)
arr5
```

```python
is_bound = (arr5 % 2 == 1) & (arr5 < 50)
arr5[is_bound]
```
