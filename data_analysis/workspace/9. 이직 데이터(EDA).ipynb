# 케글 HR 이직 데이터 셋
- IBM HR Analytics Employee Attrition & Performance 데이터셋은 기업 내 직원들의 인적 사항, 직무 만족도, 근무 환경 및 수당 등의 정보를 바탕으로 직원의 이직 여부(Attrition)를 분석하고 예측하기 위해 수집된 데이터입니다. 인사 관리자(HR)가 직원들의 이탈 원인을 파악하고 조직 만족도와 인재 유지율을 높이기 위한 비즈니스 인사이트를 도출하는 EDA 및 머신러닝 과제에 자주 활용됩니다.

## 야근(OverTime)을 하는 직원이 야근을 하지 않는 직원과 이직율의 차이가 얼마나 나는지 비교하고 결과를 서술하세요.
- 범수형 vs 범주형 데이터

```python
overtime_attrition = (
    pd.crosstab(hr_df["OverTime"], hr_df["Attrition"], normalize="index") * 100
)

print("야근 여부에 따른 이직비율(%)")
overtime_attrition
```

```python
overtime_rate = hr_df.groupby("OverTime")["Attrition"].apply(lambda x : (x == "Yes").mean() * 100)
overtime_rate
```

```python
ax = overtime_attrition.plot(kind = "bar", stacked=True)
plt.title("야근 여부에 따른 이직 비율(%)")
plt.ylabel("비율(%)")
plt.xticks(rotation =0)
plt.xlabel("야근 여부")
plt.show()

"""
    ** 범주형과 범주형 데이터를 비교할 때 ,Bar 차트가 적합
    야근을 하는 직원의 이직률이 야근을 하지않는 직원보다 약 20% 더 높다
"""
None
```

## 3번
- 가설: 마지막 승진 후 기간(YearsSinceLastPromotion)이 길어질수록 이직률이 더 높아질까?
- 검증: 데이터의 연속형, 순서형 데이터들을 시각화 -> 라인차트를 권장

```python
year_attrition = (
    pd.crosstab(hr_df["YearsSinceLastPromotion"], hr_df["Attrition"], normalize="index") * 100
)
year_attrition
```

```python
yeartime_rate = hr_df.groupby("YearsSinceLastPromotion")["Attrition"].apply(lambda x : (x == "Yes").mean() * 100)
yeartime_rate
```

```python
ax = yeartime_rate.plot(kind = "line", figsize=(10, 6))
plt.title("승진 후 시간 경과에 따른 이직 비율(%)")
plt.ylabel("비율(%)")
plt.xticks(rotation =0)
plt.xlabel("승진 후 시간")
plt.show()
```

```python
promo_df = hr_df.groupby("YearsSinceLastPromotion")["Attrition"].agg(
    YesRate=lambda x: (x == "Yes").mean() * 100,
    NoRate = lambda x : (x == "No").mean() * 100,
).reset_index()


promo_df
```

```python
# promo_df
plt.figure(figsize = (10, 8))

plt.plot(promo_df["YearsSinceLastPromotion"], promo_df["YesRate"], color = "red", label="이직(Yes)")
plt.plot(promo_df["YearsSinceLastPromotion"], promo_df["NoRate"], color = "green", label="유지(No)")

plt.title("마지막 승진 후 기간별 이직률 vs 유지율(%)")
plt.ylabel("비율(%)")
plt.xlabel("승진 후 시간")
plt.legend()
plt.axvline(3, alpha = 0.2, linestyle="--")
plt.axvline(6, alpha = 0.2, linestyle="--")
plt.axvline(9, alpha=  0.2, linestyle="--")
plt.show()


"""
    - 승진 직후 0년차에 약 18.9% 정도의 이직률을 갖는다.
    - 승진 경력의 타이틀을 확보한 직원이 몸값을 올려서 이직하지 않을까
    - 1~2년 승진을 못하면 승진을 포기하고 이직을 하는 비율도 생각보다 높게 나타났다.
    - 장기화 7년차 이후부터는 다시 이직률(0%)이 낮아지는 역잔 현상을 일어나고 있다. - 데이터 부족
    - 승진 후 기간이 길어질수록 이직률이 높아진다는 부분 기각
"""
None
```
