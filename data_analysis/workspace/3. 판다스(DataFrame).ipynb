## 1등석과 3등석의 생존율 비교

```python
first_class = titanic_df["Pclass"] == 1
survived_first = (titanic_df["Pclass"] == 1) & (titanic_df["Survived"] == 1)
third_class = titanic_df["Pclass"] == 3
survived_third = (titanic_df["Pclass"] == 3) & (titanic_df["Survived"] == 1)
```

```python
titanic_df[survived_first]
```

```python

print(f" 1등석 승객들 생존율 약 {round(len(titanic_df[survived_first]) / len(titanic_df[first_class]), 4) * 100:.2f}")
print(f" 3등석 승객들 생존율 약 {round(len(titanic_df[survived_third]) / len(titanic_df[third_class]), 4) * 100:.2f}")
```

```python
titanic_df.groupby("Pclass")["Survived"].agg(["mean"])
titanic_df.groupby("Pclass")["Survived"].mean()
```

```python
survived_pclass = titanic_df.groupby("Pclass")["Survived"].mean()
pd.DataFrame(data = survived_pclass * 100, index = [1, 2, 3]) 
```

```python
survived_pclass = titanic_df.groupby("Pclass")["Survived"].mean()
survived_pclass.index = ["1등석", "2등석", "3등석"]
survived_pclass

pd.DataFrame(data = survived_pclass * 100, index = ["1등석", "2등석", "3등석"]) 
```

## 파생변수 만들기연습

```python

#만약 NaN이라면 : 나이 알수없음
# 나이가 19세 이상 : 성인
# 19세 미만 : 아이
def get_adult(age):
    if np.isnan(age):
        return "알수없음"

    return "성인" if age>= 19 else "아이"


get_adult(20)
```

```python
np.nan == np.nan
```

```python
# .apply(): 앞의 시리즈의 값을 순차적으로 ㅂ다아서 결과를 바꾸고 새로운 결과를 반환하는 함수
titanic_df["is_adult"] = titanic_df["Age"].apply(get_adult)
```

```python
#Fare > 300 바가지 요금
#Fare <= 300: 정상요금

def rip_off(Fare):
    return "바가지 요금 " if Fare > 300 else " 정상요금 " 
```

```python
titanic_df["rip_off"] = titanic_df["Fare"].apply(rip_off)
```

```python
print(titanic_df["rip_off"])
```

```python
# 파생변수 IsAdult, RipOff 두 개를 모두 삭제
titanic_df.drop(["IsAdult","ripOff"], axis = 1, inplace = True)
```
