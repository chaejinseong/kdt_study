## 데코레이터

```python
def log_time(original_function):
    def logging(*args):
        print(datetime.now())
        return original_function(*args)
    return logging
```

```python
@log_time
def add(num1, num2):
    return num1 + num2
```

```python
add(10, 20)
```

```python
def device(num1, num2):
    return int(num2) / int(num2)
```

```python
def safe_calc(original_function):
    def exception_wrapper(*args):

        try:
            #메인 로직
            return original_function(*args)
        except ZeroDivisionError as e:
            print("0으로 나눌 수 없습니다.")
        except ValueError as e:
            print("정수만 입력할 수 있습니다")
        except TypeError as e:
            print("정수만 계산할 수 있습니다.")
        except Exception as e:
            print("알 수 없는 오류 발생")
    return exception_wrapper

```

```python
@safe_calc
def device(num1, num2):
    return int(num2) / int(num2)
```

```python
device(10, 0)
```

```python
# ketwards
# 유저 딕쳐너리를 넘기면 , 유저 정보를 모두 출력
@safe_value
def info_user(**keywards):
    print(f"{keywards["name"]}\n {keywards["age"]}\n{keywards["address"]}")
    
    
```

```python
info_user(name="홍길동")
```

```python
def safe_value(original_function):
    def exception_wrapper(*args, **keywards):
        try:
            print("args")
            original_function(*args, **keywards)
        except KeyError as e:
            print(f"{e}를 입력하세요")

    return exception_wrapper
```

```python
info_user(name="홍길동", age= "20", address="서울")
```
