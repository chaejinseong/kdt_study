```python
# 마켓(Market) 클래스
# 1) 마켓은 이름(name), 판매 상품(product), 수입(income)의 필드를 갖는다.
# 2) 판매 상품은 다음과 같다
# - 이름(name)은 “토마토”이며, 토마토의 가격(price)은 3000원, 재고(stock)은 10개이다.
# 3) 마켓은 유저에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
# - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
# - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
# - 상품 판매시 유저가 가지고 있는 할인율이 적용되어 판매가 된다.


class Market:
    def __init__(self, name, income = 0):
        self.name=  name
        self.product = {"name":"토마토", "price":3000,"stock":10}
        self.income = income
    # 3) 마켓은 유저에게 상품을 판매할 수 있는 함수(sell)을 가지고 있다.
    def sell(self, user, count):
        self.count = count
        # - 유저가 가지고 있는 돈이 상품의 판매 가격보다 적다면 판매할 수 없다.
        if user.money < self.product["price"]:
            return f"돈이 부족합니다"
        # - 유저가 구매하는 개수보다 재고의 개수가 적다면 판매할 수 없다.
        if count > self.product["stock"]:
            return f" 재고가 부족합니다"

        discounted_price =  self.product["price"] * ((100 - user.discount) / 100)
        total_price = count * discounted_price

        user.money -= total_price
        return f"구매가 완료되었습니다\n 현재 잔액 : {int(user.money)}"
        
        self.income += total_price



# 구매자(Buyer) 클래스
# 1) 구매자는 이름(name), 돈(money), 할인율(discount)의 필드를 갖는다.
class Buyer:
    def __init__(self, name, money, discount):
        self.name = name
        self.money = money
        self.discount = discount   
market = Market("Emart")
user = Buyer("hong", 100000, 20)
market.sell(user, 10)
```
