## pydantic
- 데이터 검증과 직렬화를 위한 파이썬의 라이브러리로 데이터 파이프라인 구축에 필수적으로 사용된다.
- https://pydantic.dev/docs/validation/latest/api/pydantic/base_model/

```python
#같은 데이터이고, 같은 key 값이어도 타입이 다르다.
# 클래스 
user1 = {
    "name" : "홍길동",
    "age" : 20,
    "address" : "서울시 강남"
}

cat = {
    "name" : "보고",
    "age" : 5,
    "address" : "서울"
}

class Person:
    def __init__(self, name: str, age: int, address: str):
        self.name = name
        self.age = age
        self.address = address


class Animal:
    def __init__(self, name: str, age: int, address: str):
        self.name = name
        self.age = age
        self.address = address
```

```python
my_cat = Animal(**cat)
my_hong = Person(**user1)

print(my_cat.name)
print(my_hong.name)

#타입 비교
print(my_cat)
print(my_hong)

print(isinstance(my_cat,Person))
print(isinstance(my_hong,Person))
```

```python
from pydantic import BaseModel, Field
```

```python
class Person(BaseModel):
    name: str = Field(description="사람의 이름", min_length=2, max_length=30)
    age: int = Field(description ="사람의 나이", ge=1, le=150, strict = True)
    address: str = Field(description="사람의 주소", default="알 수 없음")

#Person(name="홍길동", age="10", address="서울")
#Person(name="홍길동", age="ㄱ", address="서울")
Person(name="홍길동", age="10", address="서울")
```
