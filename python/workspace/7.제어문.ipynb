```python
# 점수 등급
# A: 90, B: 80, C: 70, D: 60, 60미만은 과락

# 사용자에게 이름과 수학, 영어, 국어 평균 점수의 등급을 매기는 로직 구현

# 출력 예시)
# 홍길동님의 총 점은 OOO점, 평균은 OO.OO점으로 OO등급입니다.
name = "cjs"
math_score = 87
English_score = 98
Korean_score = 90

grade = "A"
total = math_score+English_score+Korean_score
avg = total/ 3

if avg >= 90:
    result = f"{name}님의 총 점은 {total}점, 평균은 {avg}으로 {grade}등급 입니다."
elif avg >=80:
    grade = "B"
    result = f"{name}님의 총 점은 {total}점, 평균은 {avg}으로 {grade}등급 입니다."
elif avg >=70:
    grade = "C"
    result = f"{name}님의 총 점은 {total}점, 평균은 {avg}으로 {grade}등급 입니다."
elif avg >=60 :
    grade = "D"
    result = f"{name}님의 총 점은 {total}점, 평균은 {avg}으로 {grade}등급 입니다."

else:
    grade = "과락"
    result = f"{name}님의 총 점은 {total}점, 평균은 {avg}으로 {grade}등급 입니다."

print(result)
```
