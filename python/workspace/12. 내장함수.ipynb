```python
math_scores = [90, 85, 20, 90]
science_scores = [70, 25, 15, 90]
english_scores = [15, 100, 30, 80]
names = ["승현", "지혁", "길동", "지호"]

for student_date in zip(math_scores, science_scores, english_scores, names):
    math, science, english, name = student_date
    print(f"{name}님의 성적\n수학 점수는: {math}점\n과학 점수: {science}점\n영어 점수: {english}점", end="\n\n")
```

```python
# 짝이 맞는 개수까지만 저장된다.
math_scores = [90, 85, 20, 90]
science_scores = [70, 25, 15, 90, 10, 20, 30, 40, 50]
english_scores = [15, 100, 30, 80, 80, 80, 80]
names = ["승현", "지혁", "길동", "지호", "길동", "갈동"]

list(zip(math_scores, science_scores, english_scores, names))
```

```python
math_scores = [90, 85, 20, 90]
science_scores = [70, 25, 15, 90]
english_scores = [15, 100, 30, 80]
names = ["승현", "지혁", "길동", "지호"]

# 성적이 가장 우수한 학생 순서대로 순위 매기기(총점이 가장 높은 사람이 우수한 학생)
li1 = list(zip(math_scores, science_scores, english_scores, names))
li2 = list(map(lambda data: (data[0], data[1], data[2], data[3], data[0] + data[1] + data[2]), li1))
li3 = sorted(li2, key=lambda student: student[4], reverse=True)

for idx, student in enumerate(li3):
    print(f"{idx + 1}위 {student[3]}님, 총 점: {student[4]}점")
```
