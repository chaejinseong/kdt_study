# 변수 선언 시 주의사항
1. 되도록 선언부에 한꺼번에 선언한다(영역상단)
2. 초기화를 해준다

## 변수명 선언 시 주의 사항
1. 문자로 시작해야한다
2. 특수문자는 사용할 수 없다,(단 _ 는 허용한다)
3. 소문자로 시작한다
4. 공백을 사용할수 없다
5. 명사로 사용한다
6. 되도록 한글은 사용하지 않는다
7. 뜻이 있는 단어를 사용한다

ex)
- a, b, c, d, e, ...(x)
- data, number, age, name, ...(o)
