문제: 창고 로봇의 최단 이동
세로 N칸, 가로 M칸의 창고가 주어진다. 0은 이동 가능한 통로이고, 1은 벽이다.
로봇은 왼쪽 위 칸 (1, 1)에서 출발해 오른쪽 아래 칸 (N, M)으로 이동하려 한다. 한 번에 상하좌우로 인접한 통로 한 칸으로 이동할 수 있으며, 창고 밖으로 나갈 수 없다.
도착까지 필요한 최소 이동 횟수를 출력하라. 도착할 수 없다면 -1을 출력한다. 출발 칸에서는 이동 횟수가 0이다.
입출력 및 제한
- 첫째 줄에 정수 N, M이 주어진다. (2 ≤ N, M ≤ 500)
- 다음 N개 줄에 길이가 M인 문자열이 공백 없이 주어진다. 각 문자는 0 또는 1이다.
- 출발 칸과 도착 칸은 항상 통로이다.
- 최소 이동 횟수를 정수 하나로 출력한다.  
4 5  
00100  
10101  
00001  
11100

```python
from collections import deque
import sys

input = sys.stdin.readline

N, M = map(int, input().split())
#split() :  문자열을 나누어 리스트로 만드는 함수

warehouse = [input().strip() for _ in range(N)]
# strip() : 문자열의 양 끝의 공백 탭 줄바꿈 제거

distance = [[-1] * M for _ in range(N)]
distance[0][0] = 0
queue = deque([(0, 0)])
#deque : 양쪽 끝에서 데이터를 넣고 뺄수있는 자료구조

directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

while queue:
    row, col = queue.popleft()
    #popleft() : deque의 왼쪽 값을 꺼내어 반환하고 큐에서 제거

    if row == N - 1 and col == M - 1:
        break

    for dr, dc in directions:
        nr, nc = row + dr, col + dc

        if not (0 <= nr < N and 0 <= nc < M):
            continue

        if warehouse[nr][nc] == "1" or distance[nr][nc] != -1:
            continue

        distance[nr][nc] = distance[row][col] + 1
        queue.append((nr, nc))

print(distance[N - 1][M - 1])
```
