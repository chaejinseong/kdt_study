```python
# 심화
# 연속되는 글자의 개수 중 가장 많이 반복되는 글자 찾기
# 단, 같은 횟수라면 제일 먼저 들어오는 값을 리턴한다.
# 예상 출력결과
# text = "abcccddeaddddadwefda"
# 제일 많이 반복되는 글자 : d
# 횟수 : 4

text = "abcccddeaddddadwefda"
count =  {}

for i in text:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1
print(count)
max = 0

for j in count:
    if max < int(count[j]):
        max = count[j]
        
for key,value in count.items():
    if int(value) == max:
        print(f"제일 많이 반복되는 글자 : {key}")
        print(f"횟수 : {max}")
        
    
```

```python
text = "abcccddeaddddadwefda"

most_char = text[0]
max_count = 1

current_char = text[0]
current_count = 1

for i in range(1, len(text)):
    if text[i] == current_char:
        current_count += 1
    else: 
        current_char = text[i]
        current_count = 1

    if current_count > max_count:
        most_char = current_char
        max_count = current_count

result = f"가장 연속된 글자가 많은 문자형: {most_char}\n연속 횟수: {max_count}번"
print(result)
```
