```python
#두 수를 전달 받아서 두 수의 합을 알려주는 fhwlr
num1 = int(input("첫 번째 정수 입력:\nex)20"))
num2 = int(input("두 번째 정수 입력:\nex)20"))

result = f"두수의 합은 : {num1 + num2}"
print(result)
```

```python
#두 수를 전달 받아서 두 수의 합을 알려주는 fhwlr
num1 = float(input("첫 번째 정수 입력:\nex)20"))
num2 = float(input("두 번째 정수 입력:\nex)20"))
total = int(num1 + num2)
result = f"두수의 합은 : {total}"
print(result)
```
