```python
# 두 수를 전달하면 두수의 큰 값과 작은 값을 비교하여 결과 반환
# 치명적인 상황 - 두 수가 동일할 경우 -> early return
def numcompare(num1, num2):
    if num1 == num2:
        return "두 수가 같습니다"
    if num1 > num2:
        return f"큰수 {num1}, 작은수 {num2}"
    else:
        return f"큰수 {num2}, 작은수 {num1}"

    
```

```python
# 6) 문자열을 전달하면 글자를 뒤집어서 반환해주는 함수
# ex) "abcdef" -> "fedcba"


def stringreverse(sentence):
    count = -1
    lisen = ["" for i in range(len(sentence))]
    lisent = ""
    for i in sentence:
        lisen[count] = i
        count -= 1
    for j in lisen:
        lisent += j
    return lisent

stringreverse("abcdef")
```

```python
# 심화
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림(기본), 단 자릿수를 입력하면 자릿수대로 반올림한다.
# 사용자의 정보가 입력되면 같이 출력하기
# round()
def grade(*args, **keywords):
    avg = sum(args) / len(args)
    if "round_off" in keywords:
        avg = round(avg, keywords["round_off"])
        return f"{keywords["name"]}님의 성적 평균은 {avg} 입니다"

grade(100, 98, 85, name = "채진성", round_off = 2)
```

```python
# 심화
# 국어, 영어, 수학 점수 평균 구하기
# 자릿수는 2번째 자리부터 반올림(기본), 단 자릿수를 입력하면 자릿수대로 반올림한다.
# 사용자의 정보가 입력되면 같이 출력하기
# round()


def get_average(**keywards):
    kor, eng, math = keywards.get("kor",0),keywards.get("eng",0),keywards.get("math",0)
    total = kor + eng + math
    average = total / 3

    average = round(average, keywards.get("round", 2))
    if "name" in keywards:
        name = keywards["name"]
        return f"{name}님의 평균은 : {average}"
        
    
    return f"평균은 : {average}"
```

```python
# 두 값의 사칙연산 중 계산하고 싶은 값만 리턴해주기
# calc가 있을 때에만 계산해서 리턴, 없을 때에는 (10, 20)을 리턴

# 1. 두 개의 변수는 따로 변수에 받기
# 2. 부호는 따로 keywards 문법으로 받기

# ex) 10, 20, "+"



def cal(**calc):
    total = 0
    num1, num2 = calc.get("num1",0), calc.get("num2",0)
    if "math_sign" in calc:
        if calc["math_sign"] == "+":
            total = num1 + num2
        elif calc["math_sign"] == "-":
            total = num1 - num2
        elif  calc["math_sign"] == "*":
            total = num1 * num2
        elif  calc["math_sign"] == "/":
            total = num1 / num2
        return f"{num1} {calc["math_sign"]} {num2} = {total}"
    else:
        return f"({num1},{num2})"
```

```python
cal(num1 = 10, num2 = 12, math_sign = "+")
```

```python
# 두 값의 사칙연산 중 계산하고 싶은 값만 리턴해주기
# calc가 있을 때에만 계산해서 리턴, 없을 때에는 (10, 20)을 리턴

# 1. 두 개의 변수는 따로 변수에 받기
# 2. 부호는 따로 keywards 문법으로 받기

# ex) 10, 20, "+"
def calculator(num1, num2, **keywards):
    # print(keywards.get("calc", False))
    # print("calc" in keywards)
    
    # 부호가 없다면 튜플로 반환 (num1, num2)
    if not "calc" in keywards:
        return num1, num2

    # 부호가 있다면
    calc = keywards["calc"]
    calc_set = {"+", "-", "*", "/", "%"}
    is_confirm = calc in calc_set

    if not is_confirm:
        return "부호는 +, -, *, /, %만 사용할 수 있습니다."

    result = 0
    
    if calc == "+":
        result = num1 + num2
    elif calc == "-":
        result = num1 - num2
    elif calc == "*":
        result = num1 * num2
    elif calc == "/":
        result = num1 / num2
    elif calc == "%":
        result = num1 % num2

    return result
        
calculator(10, 20, calc="+")
```

```python
def calculator(num1, num2, **keywards):
    if not "calc" in keywards:
        return num1, num2

    calc = keywards["calc"]
    calc_set = {"+", "-", "*", "/", "%"}
    is_confirm = calc in calc_set

    if not is_confirm:
        return "부호는 +, -, *, /, %만 사용할 수 있습니다."

    return eval(f"{num1}{calc}{num2}")


calculator(10, 20, calc="+")
```
