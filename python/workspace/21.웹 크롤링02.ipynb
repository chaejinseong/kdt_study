```python
review_list = [i.text for i in naver_soup.select(".area_card .area_review_content ._text")]
reviews=zip(score_list, review_list)
```

```python
#만약 5가 있는 경우
result = []
for score, review in reviews:
    if score == 5:
        continue;

    elif score > 5:
        result.append(('추천', review))
    else:
        result.append(('비추천', review))

print(result)
```
