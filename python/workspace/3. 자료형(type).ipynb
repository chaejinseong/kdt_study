## 형변환(casting)
- 타입을 다른 타입으로 변환시키는 것

### 자동 형변환형변환

```python
# 논리 + 정수 = 정수
True + 10
```

```python
# 논리 - 정수 = 정수
False - 5
```

### 강제 형변환
- 타입을 변경하고 싶다면 변경할 타입 이름뒤에 ()를 감싼 함수에게 값을 넘긴다.
- ex) int("19"): 문자열을 정수로 변경

```python
float(10)
```

```python
bool(1)
```

```python
bool(1)
print(bool(""))
print(bool(0))
print(bool([]))
print(bool({}))
print(bool(10))
print(bool(100))
print(bool(10.5))
print(bool(True))
```

```python
int("ㄱ")
```

```python
print(int(10.5))
print(int(10.9999999999999999))
```

## type() 함수
- 값의 타입을 확인해서 결과를 반환해주는 함수

```python
print(type(10.))
```

```python
type(float(bool("hello")))
```

```python
"a" + "b"
```

```python
str((bool("h") + bool(int("1")) + bool("l") + bool("o")))
```

```python
str((bool("h")-bool("l")))
```

```python
str((bool("o")-bool("1")))
```

```python
# "h", "1", "l", "o"
# 4개의 글자를 가지고 "1004"만들기


print("1" + str((bool("h")-bool("l"))) + str((bool("o")-bool("1"))) + str((bool("h") + bool(int("1")) + bool("l") + bool("o"))))
```
