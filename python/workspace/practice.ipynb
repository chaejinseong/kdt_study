```python
# 카페 문제
# - 카페(Cafe) 클래스
# 1. 카페는 이름(name), 판매 음료(drink), 총수입(income) 필드를 가진다.
# 2. 판매 음료는 다음과 같다.
#    - 이름(name): "아메리카노"
#    - 가격(price): 4,000원
#    - 재고(stock): 15개
# 3. 카페는 손님에게 음료를 판매하는 sell() 메서드를 가진다.
#    - 손님의 돈이 할인 적용 후 결제 금액보다 적으면 판매할 수 없다.
#    - 손님이 주문한 개수보다 음료의 재고가 적으면 판매할 수 없다.
#    - 손님의 할인율을 적용한 뒤 결제한다.
#    - 할인은 음료 전체 구매 금액에 적용한다.
#    - 판매에 성공하면 손님의 돈에서 결제 금액을 차감한다.
#    - 판매에 성공하면 음료의 재고에서 구매 개수를 차감한다.
#    - 결제 금액은 카페의 income에 누적한다.
#    - 판매에 실패하면 손님의 돈, 재고, 카페의 수입은 변하지 않는다.
```

```python
class Cafe:
    def __init__(self, name, income = 0):
        self.name = name
        self.drink = {"name":"아메리카노","price":4000,"stock":15}
        self.income = income
        
    # 카페는 손님에게 음료를 판매하는 sell() 메서드를 가진다.
    def sell(self, user, count):
        self.count = count
        discounted_money = self.drink["price"]  * ((100 - user.discount) / 100)
        
         # - 손님의 돈이 할인 적용 후 결제 금액보다 적으면 판매할 수 없다.
        if user.money < discounted_money * self.count:
            return f"돈이 부족합니다"
            
        # - 손님이 주문한 개수보다 음료의 재고가 적으면 판매할 수 없다.
        if self.count > self.drink["stock"]:
            return f"재고가 부족합니다"
            
        total_money = discounted_money * self.count
        user.money -=total_money
        self.income += total_money
        self.drink["stock"] -= count
        
        


class Customer:
        def __init__(self, name, money, discount):
            self.name = name
            self.money = money
            self.discount = discount
cafe = Cafe("banapresso")
user = Customer("hong", 100000, 20)

cafe.sell(user, 10)
```
