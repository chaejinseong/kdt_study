```python
class Student:
    def __init__(self, number, name):
        self.number = number
        self.name = name

    # 클래스끼리 비교할 때 ==를 재정의
    def __eq__(self, student):
        return self.number == student.number

student1 = Student(number=1, name="홍길동")
student2 = Student(number=1, name="홍길동")

print(student1 == student2)
```

```python
class Person:
    def __init__(self, name):
        self.name = name
        print("__repr__ 호출:", self.__repr__)

    def __eq__(self, other):
        if self.__repr__()==other.__repr__():
            return True

        if not isinstance(other, Person):
            return False

        if self.name == other.name:
            return self.name == other.name

        return False

class Animal:
    def __init__(self, name):
        self.name = name
```

```python
person1 = Person("홍길동")
animal1= Animal("홍길동")

print(person1.__eq__(animal1))
```
