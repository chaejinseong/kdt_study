# 컬렉션(list, tuple, set, dict)
1. list []
    - 추가, 수정, 삭제 언제나 가능
    - [] 대괄호로 묶어주고 ,(콤마)로 값을 구분한다

2. tuple ()
   - 한 번 생성되고나면 변경이 불가능
   - () 소괄호로 묶어주고 ,(콤마)로 값을 구분한다

3. set {}
   - 순서가 없는 자료 구조(비시퀀스)
   - 중복된 값을 저장이 불가능
   - {} 중괄호로 묶어주고 ,(콤마)로 값을 구분한다

4. dict {key:value}
   - 키 + 값이 한 쌍으로 관리하며 ,(콤마)로 값을 구분한다
   - key는 중복 값을 허용하지 않으며, value는 중복값을 허용한다
   - key를 통해 value를 가져온다

## 멤버쉽 연산자(in)
- "찾을값" in객체(자료구조)

```python
"a" in str1
```

```python
#어떤 값이 실수인지 또는 정수인지 출력하기
# 단 %, int, float은 사용할 수 없다

num1 = 10
num2 = 20.24

num1_1 = num1 // 1
ans = "정수" if num1 == num1_1 else "실수"

num2_1 = num2 // 1
ans2 = "정수" if num2 == num2_1 else "실수"

print(ans)
print(ans2)
```

```python
"." in str(num1)

result = "실수" if "." in str(num1) else "정수"
print(result)
```

```python
# 선언
di1 = {}
s1 = {}
print(di1)
print(type(di1))
```

```python
di3 = {
    "a":"apple",
    "b":"banana",
    "c":"cherry"
}
print(di3)
```

```python
# 값을 조회(Read)
# 대괄호 접근법
print(di3["a"])
print(di3.get("a"))
```

```python
#d를 조회
#대괄호 접근법 없는 키 접근 -> 오류
print(di3["d"])
```

```python
#.get()으로 없는 키 접근 -> 오류가 나지 않음
print(di3.get("d","대체값"))
```

```python
di3["a"] = "alpha"
print(di3)
```

```python
di3 = {
    "a":"apple",
    "b":"banana",
    "c":"cherry",
    # 수정
    "a":"alpha"
}
print(di3)
```

```python
# .(dot) 접근법
print(list(di3.items()))
print(list(di3.keys()))
print(list(di3.values()))
```
