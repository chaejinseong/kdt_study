```python
#  조건) 
#  나이는 임의로 설정, 삼항 연산자
#  ** 중요 ** 값을 모두 변수에 담아보기

#  실습)
#  1. 나이가 19살 이상이면 "입장 가능" 출력
#  19살 미만이면 "입장 불가" 출력
#  2. 나이가 짝수면 "짝수" 출력
#  나이가 홀수면 "홀수" 출력
#  3. 나이가 16살이면 "당첨" 출력
#  아니면 "꽝입니다" 출력

#  예상 결과)
#  콘솔 결과: [본인이름]님은 [입장 가능]
#  [나이], [홀수], [꽝입니다].  

age = 999
result1 = "입장가능" if age >= 10 else "입장 불가"
result2 = "짝수" if age % 2 == 0 else "홀수"
result3 = "당첨" if age == 16 else "꽝입니다"
#print(result1)
print("콘솔 결과 : [채진성]님은", "[",result1,"]")
print("[",age,"]",",", "[",result2,"]",",", "[",result3,"]")

print("콘솔 결과 : [채진성님]은", end="")
print(" [", end="")
print(result1, end="")
print("]")
print("[",end="")
print(age,end="")
print("], ", end="")
print("[", end="")
print(result2, end="")
print("],", end="")
print(" [",end="")
print(result3, end="")
print("].")



```
