  문자열 안에서 특수한 기능을 가지는 문자들을 의미한다.

# 문자열('', "")

```python
# 함수 : 이름뒤에 소괄호, 기능 또는 값
print("저의 이름은 \"채진성\"이고, \n나이는 \"25\"살 입니다.")
```

## 규칙
- 1. 내용 안에 들어가는 내용이라면 안으로 들ㅇ여쓰기 해준다. (tab)
- 2. 내용들끼리는 같은 depth(깊이)에 들여쓰기를 사용해야 한다.  
     만약 들여쓰기를 잘못 넣었다면[Shift] + [Tab]으로 바깥으로 꺼내준다.
  3. 띄어쓰기(" ") -> 공백문자를 잘못 입력하면 depth로 인식한다.

```python
print("""
    안녕하세요
    홍길동입니다
""")
print("20")
```

```python
#출력결과를 만드시오
#출력 예상 결과  :
# ex)
# 100 * "20" * 20 = ?

print("100 * \"20\" * 20 = ?")
print("100", "*", "\"20\"", "*", "20", "=", "?")
print('100 * "20" * 20 = ?')
```
